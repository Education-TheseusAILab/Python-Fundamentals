### Exercise 3.2 (Intermediate)

**Task**: You have 47 students and want to divide them into groups of 5. Calculate how many complete groups you can make and how many students will be left over.

````python
# Solution 3.2

# Define values
total_students = 47
group_size = 5

# Calculate complete groups using floor division
complete_groups = total_students // group_size

# Calculate remaining students using modulo
remaining_students = total_students % group_size

# Print results
print(f"Total students: {total_students}")
print(f"Group size: {group_size}")
print(f"Complete groups: {complete_groups}")
print(f"Students left over: {remaining_students}")
````

**Explanation**:

- **Floor division (`//`)** divides and rounds down to the nearest integer
  - `47 // 5 = 9` (47 ÷ 5 = 9.4, rounded down to 9)
- **Modulo (`%`)** gives the remainder after division
  - `47 % 5 = 2` (because 47 = 9 × 5 + 2)
- These two operators are incredibly useful for:
  - Pagination (dividing items into pages)
  - Determining even/odd numbers (`n % 2`)
  - Cycling through items (using modulo to wrap around)

### Exercise 3.3 (Advanced)

**Task**: Calculate compound interest. Formula: A = P(1 + r)^t where P=1000 (principal), r=0.05 (5% rate), t=3 (years). Print the final amount and the interest earned.

````python
# Solution 3.3

# Define variables
principal = 1000    # Initial amount (P)
rate = 0.05         # Interest rate as decimal (r) - 5% = 0.05
time = 3            # Number of years (t)

# Calculate final amount using compound interest formula
# A = P * (1 + r)^t
final_amount = principal * (1 + rate) ** time

# Calculate interest earned
interest_earned = final_amount - principal

# Print results
print(f"Principal: ${principal}")
print(f"Annual Rate: {rate * 100}%")  # Convert back to percentage for display
print(f"Time: {time} years")
print(f"Final Amount: ${final_amount:.2f}")
print(f"Interest Earned: ${interest_earned:.2f}")
````

**Explanation**:

- **Exponentiation (`**`)** is used for powers: `(1 + rate) ** time`
- Order of operations matters:
  1. `(1 + rate)` = 1.05 (parentheses first)
  2. `1.05 ** 3` = 1.157625 (exponentiation)
  3. `1000 * 1.157625` = 1157.625 (multiplication)
- The interest rate is converted from percentage to decimal (5% = 0.05)
- `:.2f` formats to 2 decimal places for currency display

### Exercise 4.1 (Basic)

**Task**: Given two boolean variables `is_raining = True` and `has_umbrella = False`, determine if someone will get wet (they get wet if it's raining AND they don't have an umbrella).

````python
# Solution 4.1

# Given conditions
is_raining = True
has_umbrella = False

# Determine if they'll get wet
# Gets wet if: it's raining AND they DON'T have an umbrella
will_get_wet = is_raining and not has_umbrella

# Print results
print(f"Is it raining? {is_raining}")
print(f"Has umbrella? {has_umbrella}")
print(f"Will get wet? {will_get_wet}")
````

**Explanation**:

- `not has_umbrella` flips `False` to `True` (they DON'T have an umbrella)
- `is_raining and not has_umbrella` = `True and True` = `True`
- Both conditions must be true:
  - It IS raining (`True`)
  - They do NOT have an umbrella (`not False` = `True`)

### Exercise 4.2 (Intermediate)

**Task**: Create a simple eligibility checker for a loan. A person is eligible if they are at least 18 years old AND (have income >= 30000 OR have a co-signer). Test with: age=25, income=25000, has_cosigner=True.

````python
# Solution 4.2

# Given conditions
age = 25
income = 25000
has_cosigner = True

# Check eligibility
# Must be 18+ AND (income >= 30000 OR have co-signer)
is_adult = age >= 18
meets_income = income >= 30000
has_financial_backing = meets_income or has_cosigner

is_eligible = is_adult and has_financial_backing

# Print detailed breakdown
print(f"Age: {age} (Adult: {is_adult})")
print(f"Income: ${income} (Meets threshold: {meets_income})")
print(f"Has co-signer: {has_cosigner}")
print(f"Has financial backing (income OR co-signer): {has_financial_backing}")
print(f"\nLoan eligible: {is_eligible}")

# Alternative: All in one line
is_eligible_oneliner = age >= 18 and (income >= 30000 or has_cosigner)
print(f"Eligible (one-liner check): {is_eligible_oneliner}")
````

**Explanation**:

- **Parentheses are crucial** here: `(income >= 30000 or has_cosigner)` must be evaluated together
- Without parentheses, `and` has higher precedence than `or`, which would give wrong results
- Breaking down into intermediate variables (`is_adult`, `meets_income`, etc.) makes logic clearer
- The person is eligible because:
  - `age >= 18` = `True` (25 >= 18)
  - `income >= 30000` = `False` (25000 < 30000)
  - `has_cosigner` = `True`
  - `False or True` = `True` (has financial backing)
  - `True and True` = `True` (eligible!)

### Exercise 4.3 (Advanced)

**Task**: A user can access premium content if: (they are a subscriber OR they have a trial) AND (they are not banned) AND (their account is verified). Create variables for each condition and calculate the final access status.

### Exercise 5.2 (Intermediate)

**Task**: Create a simple ticket pricing system. Base price is $10. If age < 12 OR age >= 65, apply a 50% discount. If it's a weekend (is_weekend = True), add $3. Print the final price.

````python
# Solution 5.2

# Input values
age = 70
is_weekend = True

# Constants
BASE_PRICE = 10
DISCOUNT_RATE = 0.5  # 50% discount
WEEKEND_SURCHARGE = 3

# Start with base price
price = BASE_PRICE
print(f"Base price: ${price}")

# Apply age discount if eligible
if age < 12 or age >= 65:
    discount = price * DISCOUNT_RATE
    price = price - discount
    print(f"Age discount applied (50%): -${discount}")

# Add weekend surcharge if applicable
if is_weekend:
    price = price + WEEKEND_SURCHARGE
    print(f"Weekend surcharge: +${WEEKEND_SURCHARGE}")

# Print final price
print(f"\nFinal price: ${price}")
````

**Explanation**:

- We use **two separate `if` statements**, not `if-elif`, because both conditions can apply independently
- A 70-year-old on a weekend gets BOTH the discount AND the surcharge
- Using CONSTANT names (all caps) for fixed values is a Python convention for readability
- The logic flow:
  1. Start with base price ($10)
  2. Check age discount ($10 * 0.5 = $5 discount → $5)
  3. Check weekend surcharge ($5 + $3 = $8)

### Exercise 5.3 (Advanced)

**Task**: Create a password strength checker. A password is:
- "Weak" if length < 6
- "Medium" if length >= 6 but < 10 AND has at least one number
- "Strong" if length >= 10 AND has at least one number AND has at least one uppercase letter
- Otherwise "Medium"

````python
# Solution 5.3

# Test password
password = "MySecret123"  # Try: "abc", "pass123", "longpassword"

# Check characteristics
length = len(password)
has_number = any(c.isdigit() for c in password)
has_uppercase = any(c.isupper() for c in password)

# Print analysis
print(f"Password: {password}")
print(f"Length: {length}")
print(f"Has number: {has_number}")
print(f"Has uppercase: {has_uppercase}")

# Determine strength
if length < 6:
    strength = "Weak"
elif length >= 10 and has_number and has_uppercase:
    strength = "Strong"
elif length >= 6 and has_number:
    strength = "Medium"
else:
    strength = "Medium"  # Fallback for long passwords without numbers

print(f"\nPassword strength: {strength}")
````

**Explanation**:

- `len(password)` returns the number of characters
- `any(c.isdigit() for c in password)` is a **generator expression** that:
  - Iterates through each character `c` in the password
  - Checks if it's a digit with `.isdigit()`
  - `any()` returns `True` if ANY character passes the test
- Similarly, `.isupper()` checks for uppercase letters
- **Order matters** in if-elif: We check "Strong" before "Medium" because Strong passwords also meet Medium criteria
- The final `else` handles edge cases like "longpasswordnouppercase" (length >= 10 but no number)

**Task**: Write a for loop that prints the numbers 1 to 10.

````python
# Solution 6.1

# Method 1: Using range(1, 11)
print("Numbers 1 to 10:")
for i in range(1, 11):  # range(1, 11) gives 1, 2, 3, ..., 10
    print(i)

# Method 2: Print on one line
print("\nOn one line:")
for i in range(1, 11):
    print(i, end=" ")  # end=" " prevents newline, adds space instead
print()  # Final newline
````

### Exercise 6.2 (Intermediate)

**Task**: Write a program using a while loop that starts with n=1 and keeps doubling n until it exceeds 1000. Print each value of n.

````python
# Solution 6.2

n = 1
count = 0  # Optional: track iterations

print("Doubling sequence:")
while n <= 1000:  # Continue while n is 1000 or less
    print(f"  n = {n}")
    n = n * 2  # Double n (could also write: n *= 2)
    count += 1

print(f"\nFinal value (exceeded 1000): {n}")
print(f"Total iterations: {count}")
````

**Explanation**:

- `while n <= 1000` continues as long as n hasn't exceeded 1000
- Inside the loop, we print n BEFORE doubling, so we see 1, 2, 4, 8, ... 512, 1024
- Wait, why 1024? Because we check the condition BEFORE entering the loop body
- Actually, let me trace through:
  - n=1: 1 <= 1000? Yes → print 1, n becomes 2
  - n=2: 2 <= 1000? Yes → print 2, n becomes 4
  - ... continues ...
  - n=512: 512 <= 1000? Yes → print 512, n becomes 1024
  - n=1024: 1024 <= 1000? No → exit loop
- So we see 1, 2, 4, 8, 16, 32, 64, 128, 256, 512 (powers of 2 up to 1000)

### Exercise 6.3 (Advanced)

**Task**: Write a program that finds all prime numbers between 2 and 50. A prime number is only divisible by 1 and itself. Use nested loops.

````python
# Solution 6.3

print("Prime numbers between 2 and 50:")

primes = []  # Store primes in a list

for num in range(2, 51):  # Check each number from 2 to 50
    is_prime = True  # Assume it's prime until proven otherwise

    # Check if num is divisible by any number from 2 to num-1
    for divisor in range(2, num):
        if num % divisor == 0:  # Found a divisor!
            is_prime = False
            break  # No need to check further

    if is_prime:
        primes.append(num)
        print(num, end=" ")

print(f"\n\nTotal primes found: {len(primes)}")
print(f"List: {primes}")
````

**Explanation**:

- **Outer loop**: Checks each candidate number from 2 to 50
- **Inner loop**: Tests if the number is divisible by anything other than 1 and itself
- **`break`**: Exits the inner loop early when we find a divisor (no need to keep checking)
- A number is prime if we complete the inner loop without finding any divisors

**Optimization note**: We could check only up to `sqrt(num)` instead of `num-1`, but this simpler version is fine for learning.

### Exercise 7.3 (Advanced)

**Task**: Given a list of student dictionaries, find the average score, the student with the highest score, and create a list of names of students who scored above 80.

````python
# Solution 7.3

students = [
    {"name": "Alice", "score": 85},
    {"name": "Bob", "score": 72},
    {"name": "Charlie", "score": 90},
    {"name": "Diana", "score": 78}
]

# 1. Find the average score
total_score = 0
for student in students:
    total_score += student["score"]
average_score = total_score / len(students)

# Alternative using list comprehension:
# average_score = sum(s["score"] for s in students) / len(students)

print(f"Average score: {average_score}")

# 2. Find student with highest score
highest_student = students[0]  # Start with first student
for student in students:
    if student["score"] > highest_student["score"]:
        highest_student = student

# Alternative using max() with key:
# highest_student = max(students, key=lambda s: s["score"])

print(f"Highest scorer: {highest_student['name']} with {highest_student['score']}")

# 3. Names of students who scored above 80
high_scorers = []
for student in students:
    if student["score"] > 80:
        high_scorers.append(student["name"])

# Alternative using list comprehension:
# high_scorers = [s["name"] for s in students if s["score"] > 80]

print(f"Students above 80: {high_scorers}")
````

**Explanation**:

1. **Average**: Sum all scores, divide by count. Loop through and accumulate.
2. **Highest**: Keep track of the best so far, update when finding a higher score.
3. **Filter**: Check each student, add to new list if they meet the criteria.

**List comprehensions** (shown as alternatives) are a more "Pythonic" way to do these operations, but the loop versions are clearer for beginners.

### Exercise 8.2 (Intermediate)

**Task**: Write a function `is_palindrome(text)` that returns True if the text is a palindrome (reads the same forwards and backwards), ignoring case.

````python
# Solution 8.2

def is_palindrome(text):
    """Check if text is a palindrome (ignoring case).

    Args:
        text: String to check

    Returns:
        True if palindrome, False otherwise
    """
    # Convert to lowercase for case-insensitive comparison
    text_lower = text.lower()

    # Reverse the string using slicing
    reversed_text = text_lower[::-1]

    # Compare original (lowercase) with reversed
    return text_lower == reversed_text

# Test cases
test_words = ["Racecar", "hello", "Level", "Python", "A", "Madam"]

for word in test_words:
    result = is_palindrome(word)
    print(f"{word}: {'✓ Palindrome' if result else '✗ Not palindrome'}")
````

**Explanation**:

- `.lower()` converts to lowercase for case-insensitive comparison
- `[::-1]` is a **slice** that reverses a string:
  - `[start:end:step]` - empty start/end means beginning/end
  - `-1` step means go backwards
- The function returns the boolean result of the comparison

**Bonus**: For a stricter palindrome check (ignoring spaces and punctuation), you could add:

text_clean = ''.join(c for c in text.lower() if c.isalnum())

### Exercise 8.3 (Advanced)

**Task**: Write a function `analyze_text(text)` that returns a dictionary with word_count, char_count, longest_word, and word_frequency.

````python
# Solution 8.3

def analyze_text(text):
    """Analyze text and return statistics.

    Args:
        text: String to analyze

    Returns:
        Dictionary with word_count, char_count, longest_word, word_frequency
    """
    # Split text into words
    words = text.split()

    # Count words
    word_count = len(words)

    # Count characters (excluding spaces)
    char_count = len(text.replace(" ", ""))

    # Find longest word
    longest_word = ""
    for word in words:
        if len(word) > len(longest_word):
            longest_word = word
    # Alternative: longest_word = max(words, key=len) if words else ""

    # Count word frequency
    word_frequency = {}
    for word in words:
        word_lower = word.lower()  # Normalize to lowercase
        if word_lower in word_frequency:
            word_frequency[word_lower] += 1
        else:
            word_frequency[word_lower] = 1

    # Return results as dictionary
    return {
        "word_count": word_count,
        "char_count": char_count,
        "longest_word": longest_word,
        "word_frequency": word_frequency
    }

# Test the function
sample_text = "The quick brown fox jumps over the lazy dog. The fox is quick."

result = analyze_text(sample_text)

print(f"Text: \"{sample_text}\"\n")
print(f"Word count: {result['word_count']}")
print(f"Character count (no spaces): {result['char_count']}")
print(f"Longest word: {result['longest_word']}")
print(f"\nWord frequency:")
for word, count in result['word_frequency'].items():
    print(f"  '{word}': {count}")
````

**Explanation**:

- `.split()` breaks a string into a list of words (splits on whitespace by default)
- `.replace(" ", "")` removes all spaces for character counting
- For word frequency, we use a dictionary to count occurrences
- The function returns a **dictionary** with all results - a common pattern in Python for returning multiple values

**Note**: This simple version doesn't handle punctuation ("dog." vs "dog"). For production code, you'd want to strip punctuation from words.

### Exercise 8.5 (Intermediate)

**Task**: Write a recursive function that sums numbers from 1 to n.

````python
# Solution 8.5
# Recursive sum
def sum_to_n(n):
    if n == 1:
        return 1
    return n + sum_to_n(n - 1)

print("Sum to 5:", sum_to_n(5))
````

**Explanation**:

- **Recursion** is when a function calls itself
- Every recursive function needs a **base case** to stop the recursion (here: `if n == 1`)
- The function breaks down `sum_to_n(5)` into `5 + sum_to_n(4)`, then `4 + sum_to_n(3)`, etc.
- Call stack: `sum_to_n(5) → 5 + sum_to_n(4) → 5 + 4 + sum_to_n(3) → ... → 5+4+3+2+1 = 15`
- Recursion is elegant for problems with self-similar structure (trees, fractals, divide-and-conquer)

- `yield` pauses the function and saves its state; next call resumes from there
- `list()` converts the generator to a list by consuming all yielded values
- Use generators for large datasets or infinite sequences

### Exercise 8.7 (Advanced)

**Task**: Write a decorator that measures and prints execution time of a function.

````python
# Solution 8.7
import time

# Timing decorator
def timing(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(f"{func.__name__} took {(end - start)*1000:.2f} ms")
        return result
    return wrapper

@timing
def slow_task():
    total = 0
    for i in range(100000):
        total += i
    return total

slow_task()
````

**Explanation**:

- **Decorators** modify or enhance functions without changing their code
- `@timing` is syntactic sugar for `slow_task = timing(slow_task)`
- The decorator wraps the original function, adding timing logic before and after
- `*args, **kwargs` allow the wrapper to accept any arguments
- Common uses: logging, authentication, caching, timing, validation

### Exercise 8.8 (Advanced)

**Task**: Create a custom iterator that iterates over the letters of a word.

````python
# Solution 8.8
# Custom iterator over letters
class LetterIterator:
    def __init__(self, word):
        self.word = word
        self.index = 0
    def __iter__(self):
        return self
    def __next__(self):
        if self.index >= len(self.word):
            raise StopIteration
        ch = self.word[self.index]
        self.index += 1
        return ch

for ch in LetterIterator("code"):
    print(ch)
````

**Explanation**:

- **Iterators** are objects that implement `__iter__()` and `__next__()` methods
- `__iter__()` returns the iterator object (usually `self`)
- `__next__()` returns the next value or raises `StopIteration` when done
- The `for` loop automatically calls these methods behind the scenes
- Custom iterators give you control over how objects are traversed

### Exercise 9.3 (Advanced)

**Task**: Create a simple data logger with `log_entry(message)` and `read_logs()` functions.

````python
# Solution 9.3

from datetime import datetime
import os

LOG_FILE = "log.txt"

def log_entry(message):
    """Append a timestamped message to the log file.

    Args:
        message: The message to log
    """
    # Get current timestamp
    timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

    # Format the log entry
    log_line = f"[{timestamp}] {message}\n"

    # Append to file (creates file if it doesn't exist)
    with open(LOG_FILE, "a") as file:
        file.write(log_line)

    print(f"Logged: {message}")

def read_logs():
    """Read and return all log entries.

    Returns:
        List of log entries, or empty list if file doesn't exist
    """
    try:
        with open(LOG_FILE, "r") as file:
            entries = file.readlines()
            return [entry.strip() for entry in entries]  # Remove newlines
    except FileNotFoundError:
        return []

# Test the logger
print("=== Testing Logger ===\n")

# Log some messages
log_entry("Application started")
log_entry("User logged in: alice@example.com")
log_entry("API call completed successfully")

# Read logs
print("\n=== Reading Logs ===\n")
logs = read_logs()
for i, entry in enumerate(logs, 1):
    print(f"{i}. {entry}")

# Cleanup
if os.path.exists(LOG_FILE):
    os.remove(LOG_FILE)
    print("\nLog file cleaned up.")
````

**Explanation**:

- `datetime.now()` gets the current date and time
- `.strftime()` formats the datetime as a string ("2024-01-15 10:30:45")
- Mode `"a"` = append (adds to end of file, creates if doesn't exist)
- `try-except` handles the case where the log file doesn't exist yet
- `.readlines()` returns a list of lines (each including `\n`)
- `.strip()` removes whitespace/newlines from each line

This is a simplified version of what logging frameworks like Python's built-in `logging` module do.

## Section 11: Object-Oriented Programming (OOP) - Solutions


````python
# Solution 11.1
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def introduce(self):
        return f"Hi, I am {self.name} and I am {self.age} years old."

print(Person("Maya", 22).introduce())
````

**Explanation**:

- **Classes** are blueprints for creating objects with shared structure and behavior
- `__init__` is the constructor, called when creating a new instance
- `self` refers to the current instance and must be the first parameter of methods
- **Attributes** (`self.name`, `self.age`) store data specific to each instance
- **Methods** (`introduce`) define behaviors that operate on instance data

---

````python
# Solution 11.2
class BankAccount:
    def __init__(self, balance=0):
        self.balance = balance
    def deposit(self, amount):
        self.balance += amount
    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient funds")
            return
        self.balance -= amount

acct = BankAccount(100)
acct.withdraw(30)
print("Balance:", acct.balance)
acct.withdraw(100)
````

**Explanation**:

- **Default parameters** (`balance=0`) make arguments optional
- Methods can modify instance attributes (`self.balance`)
- **Encapsulation**: The class bundles data and methods that operate on that data
- **Validation**: The `withdraw` method checks for sufficient funds before proceeding
- This pattern is common for modeling real-world entities with state and behavior

---

````python
# Solution 11.3
class Student(Person):
    def __init__(self, name, age, grade):
        super().__init__(name, age)
        self.grade = grade
    def introduce(self):
        base = super().introduce()
        return f"{base} My grade is {self.grade}."

print(Student("Leo", 19, "A").introduce())
````

**Explanation**:

- **Inheritance** lets a class (`Student`) inherit from a parent class (`Person`)
- `super().__init__(...)` calls the parent's constructor to initialize inherited attributes
- The child class can add new attributes (`self.grade`) and override methods
- **Method overriding**: `Student.introduce()` replaces `Person.introduce()` but can call it via `super()`
- Inheritance models "is-a" relationships: a Student IS a Person with additional properties
