## Section 1: Introduction to Python

### What is Python?

Python is a **high-level, interpreted programming language** created by Guido van Rossum in 1991. Let's break this down:

- **High-level**: You write code that's closer to human language, not machine code
- **Interpreted**: Your code runs line by line (no separate compilation step needed)

### What Does "Interpreted" Mean?

Think of it like this:
- **Compiled languages** (C, Java): Like writing a letter, translating it fully to another language, then sending it
- **Interpreted languages** (Python): Like having a live translator who converts each sentence as you speak

This means you can:
- See results immediately
- Test small pieces of code quickly
- Debug more easily

### Why Python for AI Engineering?

1. **Rich AI/ML Ecosystem**: Libraries like TensorFlow, PyTorch, LangChain, and thousands more
2. **Readable Syntax**: Code that reads almost like English
3. **Massive Community**: Stack Overflow answers, tutorials, and help everywhere
4. **Industry Standard**: Used by Google, OpenAI, Meta, and most AI companies
5. **Rapid Prototyping**: Go from idea to working code fast

### The Python Environment

## Section 2: Data Types - The Building Blocks

### What Are Data Types?

Data types tell Python what kind of value you're working with. Just like in real life, you handle numbers differently than text.

### Why Do Data Types Matter?

- They determine what operations you can perform
- They affect how much memory is used
- They prevent errors (you can't add "hello" + 5)

### The Three Core Types We'll Cover

| Type | What It Is | Example |
|------|------------|--------|
| `int` | Whole numbers | `42`, `-7`, `0` |
| `str` | Text (strings) | `"hello"`, `'AI'` |
| `bool` | True/False | `True`, `False` |

### Integers (`int`)

Whole numbers, positive or negative, without decimals.

````python
# Creating integers
age = 25
temperature = -10
year = 2024

print("Age:", age)
print("Temperature:", temperature)
print("Year:", year)

# Check the type
print("Type of age:", type(age))
````

````python
# Python also has floats (decimal numbers)
price = 19.99
pi = 3.14159

print("Price:", price)
print("Pi:", pi)
print("Type of price:", type(price))
````

### Strings (`str`)

Text data, enclosed in quotes (single `'` or double `"` - both work!).

````python
# Creating strings
name = "Alice"
greeting = 'Hello'
message = "Welcome to AI Engineering!"

print(name)
print(greeting)
print(message)
print("Type:", type(name))
````

````python
# Strings can be combined (concatenated)
first_name = "John"
last_name = "Doe"
full_name = first_name + " " + last_name
print("Full name:", full_name)

# Strings can be repeated
print("Ha" * 3)  # HaHaHa
````

````python
# f-strings: The modern way to format strings (Python 3.6+)
# This is what you'll use most often in AI engineering

name = "Engineer"
course = "AI Engineering"
weeks = 8

# Instead of concatenation:
# message = "Hello " + name + ", welcome to " + course

# Use f-strings (note the 'f' before the quote):
message = f"Hello {name}, welcome to {course}!"
print(message)

# You can even do calculations inside f-strings
print(f"This {weeks}-week course has {weeks * 7} days of learning!")
````

### Booleans (`bool`)

The simplest type: just `True` or `False`. Essential for decision-making in code.

````python
# Creating booleans
is_enrolled = True
has_completed = False

print("Enrolled:", is_enrolled)
print("Completed:", has_completed)
print("Type:", type(is_enrolled))
````

````python
# Booleans often come from comparisons
age = 25
is_adult = age >= 18
print(f"Age {age} - Is adult? {is_adult}")

score = 85
passed = score >= 60
print(f"Score {score} - Passed? {passed}")
````

### Variables: Naming Your Data

Variables are like labeled boxes that store values.

````python
# Variable naming rules:
# 1. Must start with letter or underscore
# 2. Can contain letters, numbers, underscores
# 3. Case-sensitive (age and Age are different)
# 4. Cannot be Python keywords (if, for, while, etc.)

# Good variable names (descriptive!)
user_name = "Alice"
total_score = 95
is_active = True

# Python convention: use snake_case for variables
# this_is_snake_case (words separated by underscores)

# Variables can be reassigned
counter = 10
print("Counter:", counter)
counter = 20
print("Counter now:", counter)
````

## Section 3: Basic Arithmetic Operations

### What Are Arithmetic Operations?

Just like math class, but in code! Python can be your calculator.

### Why Learn This?

- Calculate costs, scores, metrics
- Transform data for AI models
- Implement algorithms

### Python's Arithmetic Operators

| Operator | Operation | Example | Result |
|----------|-----------|---------|--------|
| `+` | Addition | `5 + 3` | `8` |
| `-` | Subtraction | `10 - 4` | `6` |
| `*` | Multiplication | `6 * 7` | `42` |
| `/` | Division | `15 / 4` | `3.75` |
| `//` | Floor Division | `15 // 4` | `3` |
| `%` | Modulo (remainder) | `15 % 4` | `3` |
| `**` | Exponentiation | `2 ** 3` | `8` |

````python
# Basic arithmetic
print("Addition: 5 + 3 =", 5 + 3)
print("Subtraction: 10 - 4 =", 10 - 4)
print("Multiplication: 6 * 7 =", 6 * 7)
print("Division: 15 / 4 =", 15 / 4)
````

````python
# Floor division and modulo - very useful!
print("Floor Division: 15 // 4 =", 15 // 4)  # Rounds down to nearest integer
print("Modulo: 15 % 4 =", 15 % 4)  # Remainder after division

# Real-world use: checking if a number is even or odd
number = 17
remainder = number % 2
print(f"{number} % 2 = {remainder}")  # 1 means odd, 0 means even
````

````python
# Exponentiation
print("2 to the power of 3:", 2 ** 3)  # 2 * 2 * 2 = 8
print("10 squared:", 10 ** 2)  # 100
print("Square root of 16:", 16 ** 0.5)  # 4.0
````

````python
# Order of operations (PEMDAS/BODMAS applies!)
result1 = 2 + 3 * 4  # Multiplication first
result2 = (2 + 3) * 4  # Parentheses first

print("2 + 3 * 4 =", result1)  # 14
print("(2 + 3) * 4 =", result2)  # 20
````

````python
# Compound assignment operators - shorthand for common operations
score = 100
print("Initial score:", score)

score += 10  # Same as: score = score + 10
print("After +10:", score)

score -= 5  # Same as: score = score - 5
print("After -5:", score)

score *= 2  # Same as: score = score * 2
print("After *2:", score)
````

## Section 4: Logical Operations

### What Are Logical Operations?

Logical operations help you combine or modify boolean values. They're the foundation of decision-making in code.

### Why Learn This?

- Combine multiple conditions ("if user is logged in AND has permission")
- Build complex filters for data
- Control program flow

### Python's Logical Operators

| Operator | Description | Example |
|----------|-------------|--------|
| `and` | Both must be True | `True and False` → `False` |
| `or` | At least one must be True | `True or False` → `True` |
| `not` | Inverts the value | `not True` → `False` |

````python
# The 'and' operator - both conditions must be True
print("True and True =", True and True)   # True
print("True and False =", True and False)  # False
print("False and False =", False and False) # False
````

````python
# The 'or' operator - at least one condition must be True
print("True or True =", True or True)    # True
print("True or False =", True or False)   # True
print("False or False =", False or False)  # False
````

````python
# The 'not' operator - flips True to False and vice versa
print("not True =", not True)   # False
print("not False =", not False)  # True
````

### Comparison Operators

These create boolean values by comparing things:

| Operator | Meaning | Example |
|----------|---------|--------|
| `==` | Equal to | `5 == 5` → `True` |
| `!=` | Not equal to | `5 != 3` → `True` |
| `>` | Greater than | `5 > 3` → `True` |
| `<` | Less than | `5 < 3` → `False` |
| `>=` | Greater than or equal | `5 >= 5` → `True` |
| `<=` | Less than or equal | `5 <= 3` → `False` |

````python
# Comparison operators in action
age = 25
min_age = 18
max_age = 65

print(f"age ({age}) == 25:", age == 25)
print(f"age ({age}) != 30:", age != 30)
print(f"age ({age}) > min_age ({min_age}):", age > min_age)
print(f"age ({age}) >= min_age ({min_age}):", age >= min_age)
````

````python
# Combining comparisons with logical operators
age = 25
has_license = True

# Can this person rent a car? (must be 21+ AND have a license)
can_rent_car = age >= 21 and has_license
print(f"Age: {age}, Has license: {has_license}")
print(f"Can rent car: {can_rent_car}")
````

````python
# Real-world example: Access control
is_admin = False
is_owner = True
is_moderator = False

# Can delete a post if admin OR owner OR moderator
can_delete = is_admin or is_owner or is_moderator
print(f"Can delete post: {can_delete}")

# Can access admin panel only if admin AND not banned
is_banned = False
can_access_admin = is_admin and not is_banned
print(f"Can access admin panel: {can_access_admin}")
````

## Section 5: Conditional Operations (if/else)

### What Are Conditional Operations?

Conditionals let your program make decisions. "IF this is true, do this. ELSE, do that."

### Why Learn This?

- Handle different scenarios (user input, API responses)
- Implement business logic
- Control which code runs when

### The if Statement

````python
# Basic if statement
temperature = 35

if temperature > 30:
    print("It's hot outside!")  # This runs if condition is True
    print("Stay hydrated!")     # This also runs (same indentation = same block)
````

````python
# if-else: handling two cases
score = 75

if score >= 60:
    print("Congratulations! You passed!")
else:
    print("Sorry, you need to try again.")
````

````python
# if-elif-else: handling multiple cases
score = 85

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"

print(f"Score: {score} -> Grade: {grade}")
````

### Nested if Statements

Sometimes you need conditions inside conditions.

````python
# Nested if example: Authentication flow
is_logged_in = True
is_admin = False
has_permission = True

if is_logged_in:
    print("User is logged in.")

    if is_admin:
        print("Admin access: Full control granted.")
    else:
        print("Regular user access.")

        if has_permission:
            print("Permission granted for this action.")
        else:
            print("Permission denied.")
else:
    print("Please log in first.")
````

````python
# Real-world example: API response handling
status_code = 200
response_data = "Success"

if status_code == 200:
    print(f"Success! Data: {response_data}")
elif status_code == 400:
    print("Bad request - check your input")
elif status_code == 401:
    print("Unauthorized - check your API key")
elif status_code == 429:
    print("Rate limited - slow down!")
elif status_code >= 500:
    print("Server error - try again later")
else:
    print(f"Unexpected status code: {status_code}")
````

````python
# Combining conditions in if statements
age = 25
income = 50000
has_good_credit = True

if age >= 18 and income >= 30000 and has_good_credit:
    print("Loan approved!")
elif age >= 18 and income >= 30000:
    print("Loan approved with higher interest rate.")
elif age >= 18:
    print("Income too low for loan.")
else:
    print("Must be 18 or older to apply.")
````

### Ternary Operator (Conditional Expression)

A shorthand for simple if-else statements.

````python
# Traditional if-else
age = 20
if age >= 18:
    status = "adult"
else:
    status = "minor"
print(status)

# Same thing with ternary operator (one line!)
status = "adult" if age >= 18 else "minor"
print(status)
````

## Section 6: Loops - while and for

### What Are Loops?

Loops let you repeat code without copying and pasting. They're essential for processing data.

### Why Learn This?

- Process lists of data (users, API responses, documents)
- Repeat actions until a condition is met
- Automate repetitive tasks

### The while Loop

Repeats while a condition is True. Great when you don't know how many times you'll need to repeat.

````python
# Basic while loop
count = 1

while count <= 5:
    print(f"Count is: {count}")
    count += 1  # IMPORTANT: update the condition variable!

print("Loop finished!")
````

````python
# Real-world example: Retry mechanism for API calls
max_retries = 3
attempt = 1
success = False

while attempt <= max_retries and not success:
    print(f"Attempt {attempt}: Trying to connect...")

    # Simulating a success on attempt 2
    if attempt == 2:
        success = True
        print("Connection successful!")
    else:
        print("Failed. Retrying...")

    attempt += 1

if not success:
    print("All retries exhausted. Please try again later.")
````

### The for Loop

Iterates over a sequence (list, string, range, etc.). Great when you know what you're iterating over.

````python
# Looping through a range of numbers
print("Using range(5):")
for i in range(5):  # 0, 1, 2, 3, 4
    print(f"  i = {i}")

print("\nUsing range(1, 6):")
for i in range(1, 6):  # 1, 2, 3, 4, 5
    print(f"  i = {i}")

print("\nUsing range(0, 10, 2):")
for i in range(0, 10, 2):  # 0, 2, 4, 6, 8 (step of 2)
    print(f"  i = {i}")
````

````python
# Looping through a string
message = "Python"
for char in message:
    print(char)
````

````python
# Looping through a list (we'll cover lists in detail later)
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(f"I like {fruit}!")
````

### Loop Control: break and continue

````python
# break: exit the loop immediately
print("Searching for 'cherry':")
fruits = ["apple", "banana", "cherry", "date"]

for fruit in fruits:
    print(f"  Checking {fruit}...")
    if fruit == "cherry":
        print(f"  Found {fruit}! Stopping search.")
        break  # Exit the loop

print("Search complete.")
````

````python
# continue: skip the rest of this iteration, go to next
print("Printing only even numbers:")

for num in range(1, 11):
    if num % 2 != 0:  # If odd
        continue  # Skip to next iteration
    print(f"  {num} is even")
````

````python
# Nested loops
print("Multiplication table (1-3):")
for i in range(1, 4):
    for j in range(1, 4):
        print(f"  {i} x {j} = {i * j}")
    print()  # Blank line between groups
````

````python
# Multiple return values (using tuples)
def get_min_max(numbers):
    return min(numbers), max(numbers)

data = [3, 1, 4, 1, 5, 9, 2, 6]
minimum, maximum = get_min_max(data)
print(f"Min: {minimum}, Max: {maximum}")
````

````python
# Keyword arguments - name the parameters when calling
def create_user(name, age, city="Unknown"):
    return {
        "name": name,
        "age": age,
        "city": city
    }

# Positional arguments
user1 = create_user("Alice", 30, "New York")
print("User 1:", user1)

# Keyword arguments (order doesn't matter)
user2 = create_user(age=25, name="Bob", city="Boston")
print("User 2:", user2)

# Mixed (positional first, then keyword)
user3 = create_user("Charlie", city="Chicago", age=35)
print("User 3:", user3)
````

````python
# Real-world example: API cost calculator
def calculate_api_cost(tokens, model="gpt-3.5-turbo"):
    """
    Calculate the cost of an API call based on tokens used.

    Args:
        tokens: Number of tokens used
        model: The model used (affects pricing)

    Returns:
        Cost in dollars
    """
    # Pricing per 1000 tokens (example rates)
    pricing = {
        "gpt-3.5-turbo": 0.002,
        "gpt-4": 0.03,
        "gpt-4-turbo": 0.01
    }

    rate = pricing.get(model, 0.002)
    cost = (tokens / 1000) * rate
    return cost

# Calculate costs for different scenarios
print(f"10k tokens (GPT-3.5): ${calculate_api_cost(10000)}")
print(f"10k tokens (GPT-4): ${calculate_api_cost(10000, 'gpt-4')}")
print(f"10k tokens (GPT-4-turbo): ${calculate_api_cost(10000, model='gpt-4-turbo')}")
````

````python
# *args and **kwargs - flexible arguments

# *args: Accept any number of positional arguments
def sum_all(*numbers):
    return sum(numbers)

print("Sum of 1,2,3:", sum_all(1, 2, 3))
print("Sum of 1,2,3,4,5:", sum_all(1, 2, 3, 4, 5))

# **kwargs: Accept any number of keyword arguments
def print_info(**info):
    for key, value in info.items():
        print(f"{key}: {value}")

print("\nPerson info:")
print_info(name="Alice", age=30, city="New York", job="Engineer")
````

### Lambda Functions

````python
# Regular function
def square(x):
    return x ** 2

# Equivalent lambda
square_lambda = lambda x: x ** 2

print("Regular:", square(5))
print("Lambda:", square_lambda(5))

# Lambdas are often used with sort, filter, map
students = [("Alice", 85), ("Bob", 92), ("Charlie", 78)]
students_sorted = sorted(students, key=lambda x: x[1], reverse=True)
print("\nSorted by score:", students_sorted)
````

### Scope: Local, Global, and Nonlocal


````python
# Local vs global scope
count = 0

def increment():
    global count
    count += 1

increment()
print("Count:", count)

# Nonlocal example
def outer():
    total = 0
    def add_one():
        nonlocal total
        total += 1
    add_one()
    return total

print("Outer total:", outer())
````

### Recursion


````python
# Factorial using recursion
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)

print("5! =", factorial(5))
````

### Generators


````python
# Generator function using yield
def count_up_to(n):
    i = 1
    while i <= n:
        yield i
        i += 1

for value in count_up_to(5):
    print(value)
````

### Decorators


````python
# Simple decorator to log function calls
def log_call(func):
    def wrapper(*args, **kwargs):
        print(f"Calling {func.__name__}...")
        return func(*args, **kwargs)
    return wrapper

@log_call
def greet(name):
    return f"Hello, {name}!"

print(greet("Ada"))
````

### Iterators and Iterables


````python
# Using iter() and next()
items = ["a", "b", "c"]
it = iter(items)
print(next(it))
print(next(it))
print(next(it))

# Custom iterator
class Countdown:
    def __init__(self, start):
        self.current = start
    def __iter__(self):
        return self
    def __next__(self):
        if self.current <= 0:
            raise StopIteration
        value = self.current
        self.current -= 1
        return value

for n in Countdown(3):
    print(n)
````

## Section 11: Object-Oriented Programming (OOP)


### What Is OOP?


OOP helps you model real-world entities using classes and objects.


````python
# A simple class
class Robot:
    def __init__(self, name):
        self.name = name
    def greet(self):
        return f"Beep boop, I am {self.name}."

r = Robot("R2-D2")
print(r.greet())
````

````python
# Inheritance example
class TeachingRobot(Robot):
    def teach(self, topic):
        return f"{self.name} teaches {topic}."

tr = TeachingRobot("Tutor")
print(tr.greet())
print(tr.teach("Python"))
````
